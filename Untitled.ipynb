```python
import json
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas
from reportlab.lib.units import cm

# Charger les données JSON
with open("eleves.json", "r", encoding="utf-8") as f:
    eleves = json.load(f)

# Configuration PDF
width, height = A4
pdf_file = "etiquettes_eleves.pdf"
c = canvas.Canvas(pdf_file, pagesize=A4)

# Logo facultatif (décommentez si vous avez un logo local)
# c.drawImage("logo_itkan.png", x, y, width, height)

def dessiner_etiquette(eleve, x, y):
    c.setFont("Helvetica-Bold", 14)
    c.drawString(x, y, "Groupe scolaire privé Elitkane Elilmi")
    c.drawString(x, y - 20, "Accès Consultation Notes BEM/BAC Blanc 2025")

    c.setFont("Helvetica", 12)
    c.drawString(x, y - 50, f"Élève : {eleve['PRENOM']} {eleve['NOM']}")
    c.drawString(x, y - 70, f"Matricule : {eleve['MATRICULE']}")
    c.drawString(x, y - 90, f"Mot de passe : {eleve['PASSWORD']}")
    
    c.setFont("Helvetica-Oblique", 10)
    c.drawString(x, y - 110, "IMPORTANT : Veuillez conserver ces informations confidentielles.")

    c.setFont("Helvetica", 11)
    c.drawString(x, y - 140, "Comment consulter vos notes ?")
    c.setFont("Helvetica", 10)
    c.drawString(x + 15, y - 160, "1. Allez sur : https://laghoub.github.io/elitkanenotes")
    c.drawString(x + 15, y - 175, "2. Entrez votre matricule et mot de passe.")
    c.drawString(x + 15, y - 190, "3. Les notes seront disponibles à partir du 18 mai 2025.")
    c.drawString(x + 15, y - 205, "4. Les corrigés types pourront aussi être téléchargés.")
    c.drawString(x + 15, y - 230, "📧 Pour toute aide : enslaghoub@gmail.com")


# Générer 2 étiquettes par page
x_margin = 2.5 * cm
y_top = height - 3 * cm
y_bottom = height / 2 - 1.5 * cm

for i in range(0, len(eleves), 2):
    dessiner_etiquette(eleves[i], x_margin, y_top)
    if i + 1 < len(eleves):
        dessiner_etiquette(eleves[i + 1], x_margin, y_bottom)
    c.showPage()

# Sauvegarder le fichier PDF
c.save()
print(f"✅ PDF généré : {pdf_file}")
```

```python
import json

# Fichiers
input_path = "eleves.json"
output_path = "eleves_matieres4am.json"

# Dictionnaire des matières par classe
subjects_by_class = {
    "4AM": [
        "Mathématiques", "Physique", "Sciences", "Arabe", "Français", "Anglais",
        "Histoire-Géo", "sciences islamiques", "Education civique", 'Dessin', 'Sport'
    ],
    
}

# Chargement du fichier
with open(input_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# Transformation
for eleve in data:
    classe = eleve.get("CLASSE")
    if classe in subjects_by_class:
        for matiere in subjects_by_class[classe]:
            eleve[matiere] = ""

# Sauvegarde
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=4)

print("Fichier transformé enregistré sous:", output_path)
```
